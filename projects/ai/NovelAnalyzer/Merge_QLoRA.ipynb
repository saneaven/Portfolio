```python
ADAPTER_REPO   = "saneaven/gemma4-31b-novelbuds-agent-qlora"
MERGED_REPO_ID = "saneaven/gemma4-31b-novelbuds-agent-merged"
GGUF_REPO_ID   = "saneaven/gemma4-31b-novelbuds-agent-gguf"

MERGED_DIR = "/content/novelbuds-merged"
GGUF_DIR   = "/content/novelbuds-gguf"
```

```python
from google.colab import userdata
import os
os.environ["UNSLOTH_DISABLE_STATISTICS"] = "1"

os.environ["HF_TOKEN"] = userdata.get("HF_TOKEN")
```

```python
from unsloth import FastModel

ADAPTER_REPO = "saneaven/gemma4-31b-novelbuds-agent-qlora"

model, tokenizer = FastModel.from_pretrained(
    model_name   = ADAPTER_REPO,
    load_in_4bit = False,
    token        = os.environ["HF_TOKEN"],
)

print("Model loaded. Ready to merge.")
```

```python
messages = [
    {
        "role": "user",
        "content": [
            {"type": "text", "text": "Briefly describe a contemplative opening scene."}
        ],
    }
]

inputs = tokenizer.apply_chat_template(
    messages,
    tokenize              = True,
    add_generation_prompt = True,
    return_tensors        = "pt",
).to("cuda")

outputs = model.generate(inputs, max_new_tokens=150, temperature=0.7, do_sample=True)
print(tokenizer.decode(outputs[0], skip_special_tokens=False))
```

```python
model.push_to_hub_merged(
    MERGED_REPO_ID,
    tokenizer,
    save_method = "merged_16bit",
    token       = os.environ["HF_TOKEN"],
    private     = True,
)
```

```python
model.push_to_hub_gguf(
    GGUF_REPO_ID,
    tokenizer,
    quantization_method = "q4_k_m",
    token               = os.environ["HF_TOKEN"],
    private             = True,
)
```
