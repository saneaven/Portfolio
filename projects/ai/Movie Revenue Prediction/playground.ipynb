```python
from sentence_transformers import SentenceTransformer

# 1. 사용할 Sentence-BERT 모델 선택 및 로드
#    - 'all-mpnet-base-v2': 높은 성능 (추천)
#    - 'all-distilroberta-v1': 성능과 속도 균형
#    - 'all-MiniLM-L6-v2': 빠르고 가벼움
model_name = 'all-MiniLM-L6-v2'  # 예시로 MiniLM 사용 (원하는 모델로 변경 가능)
try:
    model = SentenceTransformer(model_name)
    print(f"'{model_name}' 모델을 성공적으로 로드했습니다.")
except Exception as e:
    print(f"모델 로드 중 오류 발생: {e}")
    print("인터넷 연결을 확인하거나 모델 이름이 정확한지 확인하세요.")
    exit() # 모델 로드 실패 시 종료

# 2. 벡터로 변환할 문장 정의
my_sentence = "Adventure"
# 다른 문장으로 테스트 해보세요:
# my_sentence = "Sentence embeddings can represent the meaning of a sentence."
# my_sentence = "How can I use this vector in my neural network?"

# 3. 문장을 임베딩 벡터로 변환
#    model.encode()는 단일 문장 또는 문장 리스트를 입력받아
#    NumPy 배열 형태의 임베딩 벡터 (또는 벡터 리스트)를 반환합니다.
sentence_embedding = model.encode(my_sentence)

# 4. 결과 확인
print("\n--- 결과 ---")
print(f"입력 문장: \"{my_sentence}\"")
print(f"생성된 임베딩 벡터의 차원 (Shape): {sentence_embedding.shape}")

# 임베딩 벡터는 숫자로 이루어진 NumPy 배열입니다.
# 전체 벡터를 출력하면 너무 길 수 있으므로 앞부분 일부만 출력해봅니다.
print(f"임베딩 벡터 (앞 10개 값): {sentence_embedding[:10]}")

# 생성된 벡터의 타입 확인
print(f"벡터 타입: {type(sentence_embedding)}")

# 이제 'sentence_embedding' 변수에 저장된 NumPy 배열을
# 다른 숫자 특징들과 결합하여 신경망의 입력으로 사용할 수 있습니다.
# 예: combined_features = np.concatenate((sentence_embedding, other_numerical_features))
```
